```python
import pandas as pd
import numpy as np
import os

# --- Define a sample size ---
SAMPLE_SIZE = 50000

# Helper function to downcast numeric data types to save memory 
def downcast_dataframe(df):
    print("Downcasting numeric columns to save memory...")
    for col in df.select_dtypes(include=np.number).columns:
        df[col] = pd.to_numeric(df[col], downcast='float')
    return df

print(f"Loading a sample of {SAMPLE_SIZE} rows from raw data...")

try:
    train_transaction_sample = pd.read_csv('../data/raw/train_transaction.csv').sample(n=SAMPLE_SIZE, random_state=42)
    train_identity_sample = pd.read_csv('../data/raw/train_identity.csv').sample(n=SAMPLE_SIZE, random_state=42)
except FileNotFoundError as e:
    print(f"Error: {e}. Please ensure the raw data files are in the ../data/raw/ directory.")
    exit()

# Downcast the sampled dataframes to optimize memory usage
train_transaction_sample = downcast_dataframe(train_transaction_sample)
train_identity_sample = downcast_dataframe(train_identity_sample)

#Save the individual sample files
output_dir = '../data/processed/'
os.makedirs(output_dir, exist_ok=True)
train_transaction_sample.to_csv(os.path.join(output_dir, 'train_transaction_sample.csv'), index=False)
train_identity_sample.to_csv(os.path.join(output_dir, 'train_identity_sample.csv'), index=False)

print(f"\nClean processed sample files with {train_transaction_sample.shape[0]} rows saved to {output_dir}")
```
